```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv(file_path)
```

```python
# Set the overall window size for the plots
plt.figure(figsize=(12,6))

# Subplot 1: TV vs Sales
plt.subplot(1,3,1)
plt.scatter(data['TV'], data['Sales'], color='blue')
plt.title('TV Advertising vs. Sales')
plt.xlabel('TV Advertising')
plt.ylabel('Sales')

# Subplot 2: Radio vs Sales
plt.subplot(1,3,2)
plt.scatter(data['Radio'], data['Sales'], color='green')
plt.title('Radio Advertising vs. Sales')
plt.xlabel('Radio Advertising')
plt.ylabel('Sales')

# Subplot 3: Newspaper vs Sales
plt.subplot(1,3,3)
plt.scatter(data['Newspaper'], data['Sales'], color='red')
plt.title('Newspaper Advertising vs. Sales')
plt.xlabel('Newspaper Advertising')
plt.ylabel('Sales')

plt.show()
```

# Model Building

```python
# Import the function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import the Linear Regression model class (Fixed typo: LienarRegression -> LinearRegression)
from sklearn.linear_model import LinearRegression

# Import the metric to evaluate the model's performance
from sklearn.metrics import mean_squared_error
```

```python
X = data[['Newspaper']]
y = data['Sales']
```

```python
# Split the dataset into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=34)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

# Model Prediction

```python
# Make predictions on the test dataset using the trained model
y_pred = model.predict(X_test)
```

# MSE Calculation

```python
# Calculate the Mean Squared Error (MSE) using test labels and predictions
mse = mean_squared_error(y_test, y_pred)

# Print  the final MSE score to the console
print("MSE error score: ", mse)
```

# Coefficients and Intercept

```python
# Print the model coefficients (weigs for each feature)
print("Coefficient: ", model.coef_)

# Print the model intercept(bias/constant value)
print("Intercept: ", model.intercept_)
```

# Model Regression Plot

```python
# Set the window size for the final regression plot
plt.figure(figsize=(8,6))

# Plot the actual data points as blue dots
plt.scatter(X, y, color='blue', label='Actual Data')

# Plot the predicted regression line in red using the test data
plt.plot(X_test, y_pred, color='red', linewidth=2, label='Regression Line')

#Add the main title for the regression model
plt.title('Simple Linear Regression')

#Label the horizontal axis (X-axis)
plt.xlabel('TV Advertising Spend')

#Label the vertical axis (Y-axis)
plt.ylabel('Sales')

#Display the legend to show 'Actual Data' and 'Regression Line' labels
plt.legend()

#Render and output the final plot
plt.show()
```
