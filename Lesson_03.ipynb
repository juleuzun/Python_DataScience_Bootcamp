# Try - Except - Finally

```python
try:
  a = float(input("Dividend: "))
  b = float(input("Divisor: "))
  print(a/b)

except ValueError:
  print("Please enter a valid number.")

except ZeroDivisionError:
  print("If the divisor is 0, it will approach infinity. Mathematically, it is undefined.")

except KeyboardInterrupt:
    print("Exiting.")

except:
  print("an error occurred")

finally:
  print("Tech Istanbul Bootcamp")
```

# Mathematical Functions

```python
# abs - absolute value
a = -56
print(abs(a))
```

```python
# divmod - takes two numbers and returns a tuple containing their quotient and remainder.
a = 99
b = 32
c = int(a/b)
d = a/b - c
d = d*b
print(c)
print(d)

print(divmod(a,b))
```

```python
a = [10,2,6,22,40,30,100,6]
max_so_far = 0
for num in a:
  if max_so_far < num:
    max_so_far = num

print(max_so_far)
print(max(a))
print(min(a))
```

```python
total = 0
for num in a:
  total += num
print(total)
print(sum(a))
```

```python
#round -  rounds a floating-point number to a specified number of decimals.
a = 1
b = 3
c = a/b
d = 2.257777
print(round(c,3))
print(round(d,3))
```
