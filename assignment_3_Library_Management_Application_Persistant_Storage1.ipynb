```python
import os

FILE_NAME = os.path.join("/content/drive/MyDrive/Tech Istanbul", "library.txt")
books = []

# Automatically creates an empty file if it does not exist
if not os.path.exists(FILE_NAME):
  with open(FILE_NAME, "w", encoding="utf-8") as file:
    pass

# Read the file safely
with open(FILE_NAME, "r", encoding="utf-8") as file:
  for line in file:
    line = line.strip()
    if line != "":
      book_name, author_name = line.split(";")
      books.append([book_name, author_name])

print("File read successfully. Total books loaded:", len(books))
```

```python
# ==============================
# File and Book List
# ==============================

# ==============================
# File Path Configuration
# ==============================

import os

# Use the notebook's directory by default.
# This makes the project portable when it is cloned from Github.

FILE_NAME = os.path.join(os.getcwd(), "library.txt")

books = []

with open(FILE_NAME, "r", encoding="utf-8") as file:

  for line in file:

    line = line.strip()

    if line !="":

      book_name, author_name = line.split(";")
      books.append([book_name, author_name])
```

```python
# ==============================
# Main Menu
# ==============================

while True:

  print("\n===LIBRARY MANAGEMENT APPLICATION===")
  print("1 - Add Book")
  print("2 - Remove Book")
  print("3 - List Books")
  print("4 - Exit")

  choice = input("Enter your choice: ").strip()


# ==============================
# 1 - Add Book
# ==============================

  if choice == "1":
    book_name = input("Book title: ").strip()
    author_name = input("Author name: ").strip()

    found = False
    for book in books:
      if(book[0].lower() == book_name.lower() and book[1].lower() == author_name.lower()):
        found = True
        break

    if found:
      print("This book is already in the library.")
    else:
      books.append([book_name, author_name])

      #Append the book to the end of the file

      with open(FILE_NAME, "a", encoding="utf-8") as file:
        file.write(f"{book_name};{author_name}\n")

      print(f'"{book_name}"  has been added to the library.')



# ==============================
# 2 - Remove Book
# ==============================

  elif choice == "2":
    author = input("Author name: ").strip()

    found_books = []
    indexes = []

    for i in range(len(books)):
      if books[i][1].lower() == author.lower():
        found_books.append(books[i])
        indexes.append(i)

    if len(found_books) == 0:
      print("No books by this author were found.")
    else:
      print("\nBooks by this author:")
      for i in range(len(found_books)):
          print(f"{i + 1} - {found_books[i][0]}")

      try:
        selected = int(input("Enter the number of the book you want to remove: "))
      except ValueError:
        print("Invalid input. Please enter a number.")
        continue

      if (selected < 1 or selected > len(found_books)):
        print("Invalid selection.")
      else:
        actual_index = indexes[selected - 1]
        removed_book = books.pop(actual_index)

        #Rewrite the file with the remaining books
        with open(FILE_NAME, "w", encoding="utf-8") as file:
          for book in books:
            file.write(f"{book[0]};{book[1]}\n")

        print(f'You removed "{removed_book[0]}". Enjoy reading.')


# ==============================
# List Books
# ==============================

  elif choice == "3":
    if len(books) == 0:
      print("There are no books currently registered in the library.")
    else:
      print("\nBooks in the library:\n")
      for i in range(len(books)):
        print(f"{i + 1} - {books[i][0]} - {books[i][1]}")


# ==============================
# Exit
# ==============================

  elif choice == "4":
    print("Program terminated.")
    break


# ==============================
# Invalid Selection
# ==============================

  else:
    print("Invalid selection, please try again.")
```
